# How do I find the most similar pair of time series sub-sequences (motifs)?

*Reproducing Slide 3 from Eamonn Keogh's "100 Time Series Data Mining Questions (with Answers)" using STUMPY.*

```python
import matplotlib.pyplot as plt
import numpy as np
import stumpy

# 1. Load exact Slide 3 dataset (Sony AIBO Robot Dog Accelerometer & Carpet Query)
T = np.loadtxt("https://raw.githubusercontent.com/TDAmeritrade/stumpy/main/docs/Tutorial_Pattern_Matching_steam_gen.txt")
Q = np.loadtxt("https://raw.githubusercontent.com/TDAmeritrade/stumpy/main/docs/Tutorial_Pattern_Matching_carpet_query.txt")

# 2. Find closest query match in 1 line using MASS
distance_profile = stumpy.mass(Q, T)
idx = np.argmin(distance_profile)

# 3. Plot full time series with red dashed match boundaries & query overlay
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 6))
ax1.plot(T, color="black", alpha=0.8)
ax1.axvline(x=idx, color="red", linestyle="--")
ax1.axvline(x=idx + len(Q), color="red", linestyle="--")
ax1.set_ylabel("Acceleration")

ax2.plot(stumpy.core.z_norm(Q), label="Query (Carpet Pattern)", color="tab:blue")
ax2.plot(stumpy.core.z_norm(T[idx : idx + len(Q)]), label="Best Match", color="tab:orange", linestyle="--")
ax2.legend()
plt.tight_layout()
plt.show()
```

### Summary
The two red dashed lines highlight the starting positions of the most similar pair of sub-sequences (motifs) of length $m = 50$.
